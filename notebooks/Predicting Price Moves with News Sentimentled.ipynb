```python
#Load and Preprocess Historical Stock Data
import pandas as pd
import os

# Define the directory path
data_path = r'C:\Users\ayedr\week-1\data\yfinance_data'

# List of stock tickers
tickers = ['AAPL', 'AMZN', 'GOOG', 'MSFT', 'NVDA', 'TSLA']

# Dictionary to hold DataFrames
stock_data = {}

# Load each Excel file
for ticker in tickers:
    file_path = os.path.join(data_path, f'{ticker}_historical_data.xlsx')
    df = pd.read_excel(file_path)
    df['Date'] = pd.to_datetime(df['Date'])
    df['Ticker'] = ticker
    df.sort_values('Date', inplace=True)
    df['Target'] = (df['Close'].shift(-1) > df['Close']).astype(int)  # Binary price movement
    stock_data[ticker] = df
#Concatenate into a Unified Dataset
# Combine all tickers into a single DataFrame
historical_df = pd.concat(stock_data.values(), ignore_index=True)

```
